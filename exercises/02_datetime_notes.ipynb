```python
import pandas as pd
from datetime import datetime, date, time, timedelta

start_date = datetime.today()
date_start = pd.date_range(start=start_date, periods=10, freq='D')
date_end = pd.date_range(start=start_date, periods=10, freq='D')

df = pd.DataFrame()
df['Start_day'] = date_start
df['End_day'] = date_end
df['Date_diff_days'] = date_end - date_start
df['Day_of_end_date'] = df['End_day'].dt.day
df['Month_of_start_date'] = df['Start_day'].dt.month_name()
df['Year_of_start_date'] = df['Start_day'].dt.year

df.head()
```
