middel char(s)

```python
def middle(s):
    mid = len(s) // 2
    if len(s) % 2 == False:
        return s[mid-1:mid+1]
    else:
        return s[mid]

print(middle("abba"))
print(middle("hello"))
print(middle("python"))
```

Title-case safely

```python
def safe_title(s):
    for word in s.split():
        if word.isupper():
            print(word)

safe_title("learn SQL now")
safe_title("use GPU")
safe_title("hello world") 
```

```python
def safe_title(s):
    result = []
    for word in s.split():
        if word.isupper():
            result.append(word)
        else:
            result.append(word.capitalize())  
    return " ".join(result)

print(safe_title("learn SQL now"))   
print(safe_title("use GPU"))         
print(safe_title("hello world")) 
```

Reverse words

```python
def reverse_words(s):
    words = [x.strip() for x in s.split()]
    words.reverse()
    return " ".join(words)

print(reverse_words("one two  three"))
print(reverse_words("bed nine late"))
print(reverse_words("dog ball tree"))
```

```python
def reverse_words(s):
    return " ".join(s.split()[::-1])

print(reverse_words("one two  three"))
print(reverse_words("bed nine late"))
print(reverse_words("dog ball tree"))
```

Normalize spaces

```python
def squash_spaces(s):
    return " ".join(s.split())

print(squash_spaces("  a\t b   c "))
print(squash_spaces("hello   world"))
print(squash_spaces("  too   many   spaces  "))
```

CSV line -> list (no quotes)

```python
def split_csv(s):
    return [x.strip() 
            for x in s.split(',')
                if x.strip()]


print(split_csv("one, two, three"))
print(split_csv("one, two, three,"))
print(split_csv("a,  b,c,   d"))
```

Mask secret

```python
def mask_email(s):
    user, domain = s.split('@')
    domain_name, tld = s.rsplit('.', 1)
    masked_user = user[0] + '*' * (len(user)-1)
    masked_domain = domain_name[0] + '*' * (len(domain_name)-1)
    return f"{masked_user}@{masked_domain}.{tld}"

print(mask_email("alice@example.com"))
print(mask_email("bob@mail.org"))
print(mask_email("user@site.co.uk"))
```

Find all indexes

```python
def find_all(s, sub):
    result = []
    for i in range(len(s)):
        if s[i:i+len(sub)] == sub:
            result.append(i)
    return result

print(find_all("aaaa", "aa"))
print(find_all("hello", "l"))
print(find_all("banana", "an"))
```

Anagram check

```python
def is_anagram(a, b):
    a = a.lower().replace(" ", "")
    b = b.lower().replace(" ", "")
    return sorted(a) == sorted(b)

print(is_anagram("Listen", "Silent"))
print(is_anagram("hello", "world"))
print(is_anagram("Astronomer", "Moon starer"))
```

Format table row

```python
def fmt_row(values, widths):
    result = []
    for i in range(len(values)):
        result.append(values[i].ljust(widths[i]))
    return " | ".join(result)

print(fmt_row(["name", "age", "city"], [10, 5, 8]))
print(fmt_row(["alice", "30", "berlin"], [10, 5, 8]))
print(fmt_row(["bob", "25", "munich"], [10, 5, 8]))
```

Challenge - Slugify

```python
import re
def slugify(title):
    title = title.lower().strip()
    title = re.sub(r'[^a-z0-9]+', '-', title)
    return title.strip('-')


print(slugify("Hello World"))
print(slugify("  My Blog Post! "))
print(slugify("Python & Data Engineering 101"))
```
