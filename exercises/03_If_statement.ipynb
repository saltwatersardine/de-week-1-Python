1. FizzBuzz (classic)

```python
n = int(input("Pleae provide a number: "))

def fizzbuzz(n):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 5 == 0:
        print("Buzz")
    elif n % 3 == 0:
        print ("Fizz")
    else:
        print(n)

fizzbuzz(n)
```

4. Grade boundaries

```python
def grade(score):
    if score > 100 or score < 0:
        print("ValueError")
    elif score >= 90:
        print("A")
    elif score >= 70:
        print("B")
    elif score >= 50:
        print("C")
    elif score >= 30:
        print("D")
    else: 
        print("F")

grade(-2)
```

5. Password strength (simple)

```python
s = input("What is your password?: ")

def password_strength(s):
    has_letter = any(char.isalpha() for char in s)
    has_digit = any(char.isdigit() for char in s)
    has_special = any(not char.isalnum() for char in s)

    if len(s) < 6 or not has_letter or not has_digit:
        return "weak"
    elif len(s) >= 8 and has_letter and has_digit and has_special:
        return "strong"
    else:
        return "medium"

password_strength(s)
```

6. Shipping rules

```python
def shipping_cost(weight_kg, country, premium=False, local_country="IN"):
    if weight_kg <= 1:
        cost = 5
    elif weight_kg <= 5:
        cost = 10
    else:
        cost = 20

    if country != local_country:
        cost += 15

    if premium:
        cost += 25

    return cost

while True:
    try:
        weight_kg = float(input("Enter weight in kg: "))
        break
    except ValueError:
        print("Please enter a number only, try again")

while True:
    premium_input = input("Do you want premium shipping? (yes/no): ")
    if premium_input.lower() in ["yes", "no"]:
        premium = premium_input.lower() == "yes"
        break
    print("Please enter yes or no, try again")

while True:
    country = input("Enter country code: ")
    if country.isalpha():
        break
    print("Country code must be letters only, try again")

print(shipping_cost(weight_kg, country, premium))

     
```

8. Safe cast

```python
def to_int(s, default=None):
    stripped = s.strip()
    if stripped.isdigit():
        return int(stripped)
    else:
        return default



print(to_int("12"))        # 12
print(to_int(" 12 "))      # 12  (strips whitespace)
print(to_int("12.0"))      # None (not a pure integer)
print(to_int("abc"))       # None (not a number)
print(to_int("abc", 0))    # 0   (custom default)
print(to_int("", 99))      # 99  (empty string, custom default)

```

9. Guard clauses

```python
def discount(price, code):
    if price <= 0:
        return 0
    elif code == "NONE":
        return price
    elif code == "HALF":
        return price * 0.5
    else:
        return price * 0.9

print(discount(100, "NONE"))   
print(discount(100, "HALF")) 
print(discount(100, "SALE"))   
print(discount(-10, "HALF"))  
```

10. Ternary map

```python
def label_numbers(nums):
    number = []
    for n in nums:
        if n < 0:
            number.append("neg")
        elif n==0:
            number.append("zero")
        else:
            number.append("pos")
    return number


print(label_numbers([1, -2, 0, 5, -3]))   # ['pos', 'neg', 'zero', 'pos', 'neg']
print(label_numbers([0, 0, 0]))            # ['zero', 'zero', 'zero']
print(label_numbers([-1, -2, -3]))         # ['neg', 'neg', 'neg']
print(label_numbers([1, 2, 3]))            # ['pos', 'pos', 'pos']
print(label_numbers([]))                   # []
```
