Type explorer

```python
def num_type(x):
    if isinstance(x, bool):
        return "bool"
    elif isinstance (x, int):
        return "int"
    elif isinstance(x, float):
        return "float" 

print(num_type(5))
print(num_type(2.0))
print(num_type(True))
```

Safe division

```python
def safe_div(a, b, default=None):
    if b==0:
        return default
    return float(a/b)

print(safe_div(9, 3))
print(safe_div(9, 0))
print(safe_div(9, 0, -1))
```

Integer vs. float division

```python
def divisions(a, b):
    return (a // b, a / b, a % b)

print(divisions(7, 2))
print(divisions(-7, 2))
print(divisions(10, 3))
```

Rounding modes

```python
import math

def bankers_round(x, ndigits=0):
    return round(x, ndigits)        

def classic_round(x, ndigits=0):
    factor = 10 ** ndigits
    return math.copysign(math.floor(abs(x) * factor + 0.5) / factor, x)

print(bankers_round(2.5))
print(bankers_round(3.5))
print(classic_round(2.5))
print(classic_round(3.5))
print(classic_round(-2.5))
```

Parse numbers

```python
def parse_int(s, base=10, default=None):
    try:
        return int(s.strip().replace("_", ""), base)
    except ValueError:
        return default

print(parse_int("  42  "))
print(parse_int("1_000"))
print(parse_int("ff", base=16))
print(parse_int("abc"))
```

Boolean arithmetic 

```python
def true_ratio(seq):
    return sum(bool(x) for x in seq) / len(seq)

print(true_ratio([1, 0, 1, 1, 0]))
print(true_ratio([True, False, 1, 0, "", "x"]))
print(true_ratio([None, 0, "", [], 1]))
```

Min/max & aggregates 

```python
def stats(nums):
    if not nums:
        return {"count": 0, "total": 0, "mean": None, "minimum": None, "maximum": None}
    return {
        "count": len(nums),
        "total": sum(nums),
        "mean": sum(nums) / len(nums),
        "minimum": min(nums),
        "maximum": max(nums)
    }

print(stats([1, 2, 3]))
print(stats([]))
print(stats([5]))
```

Bucket by stage

```python
def bucket(x, step):
    return (x // step) * step

print(bucket(37, 10))
print(bucket(-3, 5))
print(bucket(100, 25))
```

Time math (seconds -> h:m:s)

```python
def to_hms(seconds):
    h = seconds // 3600
    m = (seconds % 3600) // 60
    s = seconds % 60
    return f"{h:02}:{m:02}:{s:02}"

print(to_hms(3661))   
print(to_hms(0))       
print(to_hms(3600))   
print(to_hms(86399))   
```

Challenge: Running totals with precise rounding

```python
def running_totals(amounts, ndigits=2):
    totals = []
    total = 0
    for amount in amounts:
        total = round(total + amount, ndigits)
        totals.append(total)
    return totals

print(running_totals([10.005, 0.005], 2)) 
print(running_totals([1.5, 2.5, 3.5], 0))  
print(running_totals([100, 50.75, 25.005], 2))  
```
